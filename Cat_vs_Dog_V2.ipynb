# Cat VS Dog

```python
from tensorflow.keras import layers, models
```

```python
model = models.Sequential()

model.add(layers.RandomFlip("horizontal", input_shape=(150, 150, 3)))
model.add(layers.RandomRotation(0.1))
model.add(layers.RandomZoom(0.1))

model.add(layers.Rescaling(1./255))

model.add(layers.Conv2D(32, (3,3), activation='relu'))
model.add(layers.MaxPooling2D((2,2)))

# Layer 2: Find simple shapes (circles, triangles)
model.add(layers.Conv2D(64, (3,3), activation='relu'))
model.add(layers.MaxPooling2D((2,2)))

# Layer 3: Find complex features (paws, ears, noses)
model.add(layers.Conv2D(128, (3,3), activation='relu'))
model.add(layers.MaxPooling2D((2,2)))

# The Brain
model.add(layers.Flatten())
model.add(layers.Dropout(0.5))
model.add(layers.Dense(128, activation='relu')) 
model.add(layers.Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer = 'adam', metrics=['accuracy'])
```

```python
from tensorflow.keras.utils import image_dataset_from_directory
```

```python
print("Loading training data...")
train_dataset = image_dataset_from_directory(
    'training_set',
    image_size=(150, 150),
    batch_size=32,
    label_mode='binary'
)

print("\nLoading test data...")
test_dataset = image_dataset_from_directory(
    'test_set',
    image_size=(150, 150),
    batch_size=32,
    label_mode='binary'
)
```

```python
print("Starting training...")

history = model.fit(
    train_dataset,
    validation_data=test_dataset,
    epochs=5
)
```

```python
import os 
import numpy as np
from tensorflow.keras.preprocessing import image

folder_path = 'final_test_set'

for file in os.listdir(folder_path):
    img_path = os.path.join(folder_path, file)

    if not file.lower().endswith(('.png', '.jpg', '.jpeg')):
        continue

    img = image.load_img(img_path, target_size=(150, 150))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)

    prediction = model.predict(img_array)

    if prediction[0][0] < 0.5:
        print(f"{file} → CAT 🐱 (Score: {prediction[0][0]:.4f})")
    else:
        print(f"{file} → DOG 🐶 (Score: {prediction[0][0]:.4f})")
```
